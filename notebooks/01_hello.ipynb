```python
def calc_EURUSD(pos,entry_price,exit_price,lot_size):
    entry_price = entry_price * 10000 #Pips calculated
    exit_price = exit_price * 10000
    if pos == 1: # 1 is buy, 0 is sell
        pips = exit_price - entry_price
    else:
        pips = entry_price - exit_price
    profit = pips * 10 * lot_size #calculate the profit
    return profit
print(calc_EURUSD(1,1.0850,1.0875,0.5))
```
