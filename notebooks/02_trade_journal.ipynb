```python
def pip_size(pair):
    return 0.01 if pair.endswith("JPY") else 0.0001
```

```python
def calc_fx_pair(pair,directions,entry_price,exit_price,lot_size):
    pips = directions * (exit_price - entry_price) / pip_size(pair)
    if pair.startswith("USD"):
        pip_value = pip_size(pair)  * 100000  / exit_price
    else :
        pip_value = pip_size(pair) * 100000
    profit = pips * pip_value * lot_size #calculate the profit
    return profit
```

```python
#Convert direction to +1 for Buy and -1 for sell
for trade in new_trades:
    if trade["dir"] == "BUY":
        trade["sign"] = 1
    elif trade["dir"] == "SELL":
        trade["sign"] = -1
```

```python
#Pnl through trade
result_profit = []
for t in new_trades:
    result_profit.append(calc_fx_pair(t["pair"], t["sign"], t["entry"], t["exit"], t["lot"]))
print(result_profit)
```

```python
#Average win/loss and winrate
total_loss = 0
total_win = 0
win_count = 0
loss_count = 0
#Calc average win/loss
for p in result_profit:
    if p > 0:
        win_count += 1
        total_win += p
    elif p < 0:
        loss_count += 1
        total_loss += p
avg_win = total_win / win_count
avg_loss = total_loss / loss_count
trade_winrate = win_count / len(result_profit) * 100
print(f"Winrate là {trade_winrate}%, trong đó Lợi nhuận trung bình: {avg_win} và lỗ trung bình là {avg_loss}")
```

```python
#R:R and expectancy
losing_streak_array = []
losing_streak = 0
expectancy = 0
risk_to_reward = avg_win / abs(avg_loss)
expectancy = sum(result_profit) / len(result_profit)
print(f"risk to reward is: {risk_to_reward} and expectancy is: {expectancy} per trade") 
for p in result_profit:
    if p < 0:
        losing_streak += 1
    else:
        losing_streak_array.append(losing_streak)
        losing_streak = 0
losing_streak_array.append(losing_streak)
print(max(losing_streak_array))
```
